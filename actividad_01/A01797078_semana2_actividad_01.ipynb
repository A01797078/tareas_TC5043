*   **Pregunta 2.**  

Busca e imprime todas las palabras que terminan con dos o más signos de admiración seguidos, por ejemplo "!!!".

Debes imprimir tanto la palabra como la totalidad de signos de admiración que le siguen.

Indica cuántos resultados obtuviste.



```python
# Usamos:
#  '\w' para buscar caracteres de palabras en Unicode
#  '!{2,}' para buscar al menos dos ocurrencias del caracter '!'
results = [res for doc in docs if (res := re.findall(r"\w+!{2,}", doc))]
```

```python
# Usamos concatenate para aplanar el arreglo en caso de
# que hubiera múltiples palabras
np_res = np.concatenate(results)
print(f"total: {len(np_res)}")
np_res
```

*   **Pregunta 3.**

```python
results = [res for doc in docs if (res := re.findall(r"\b[A-Z]+\b", doc))]
```

```python
np_res = np.concatenate(results)
print(f"total de palabras en mayúsculas: {len(np_res)}")
np_res
```

*   **Pregunta 4.**  

Busca e imprime los comentarios en donde todos los caracteres alfabéticos (letras) están en mayúsculas.

Cada coincidencia encontrada debe ser todo el comentario/enunciado.

Indica cuántos resultados obtuviste.


```python
# ^ y $: Aseguran que se regrese todo el enunciado no solo una parte.
#   A-Z: Define las letras mayúsculas de la A a la Z
# Además de las mayúsculas permitiremos los demás caractéres:
#   \d: cualquier dígito
#   \W: cualquier caracter no alfanuméricos (incluido el espacio)
results = [res for doc in docs if (res := re.findall(r"^[A-Z\d\W]+$", doc))]
```

```python
print(f"total de oraciones en mayúsculas: {len(results)}")
results
```

*   **Pregunta 5.**  

Busca e imprime todas las palabras que tengan una vocal acentuada, del tipo á, é, í, ó, ú.

Indica cuántos resultados obtuviste.

```python
# Agregamos el flag IGNORECASE para no tener que repetir
# la letra acentuada en mayúscula dentro del regex
# pero igualmente las encontrarrá
results = [res for doc in docs if (res := re.findall(r"\w*[áéíóú]\w*", doc, re.IGNORECASE))]
```

*   **Pregunta 6.**  

Busca e imprime todas las cantidades numéricas monetarias, enteras o con decimales, que inician con el símbolo $\$$.

Indica cuántos resultados obtuviste.

```python
# Usamos:
# \$ para detectar el caracter '$'
# \s* por si hubieran espacios entre el signo de dólares y el primer digito
# \d+ para detectar la parte entera
# (?:\.\d+)? para detectar la parte decimal
#   agregamos `?:` para que no se capture ese grupo
#   agregamos `?` porque la parte decimal puede estar o no incluida
results = [res for doc in docs if (res := re.findall(r"\$\s*\d+(?:\.\d+)?", doc))]
```

```python
np_res = np.concatenate(results)
print(f"total de cantidades monetarias: {len(np_res)}")
np_res
```

*   **Pregunta 7.**  

Busca e imprime todas las palabras que sean variantes de la palabra "love", sin importar si incluyen mayúsculas o minúsculas, o la manera en que esté conjugada o alguna otra variación que se haga con dicha palabra.

Indica cuántos resultados obtuviste.

```python
# Usamos re.IGNORECASE para simplificar la expresión regular
# Revisando el documento manualmente, la expresión usada captura las variaciones
#   de `love`, pero igual es muy ingenuo, porque existe la palabra `lovebird`,
#   que empieza con `love` pero no esta relacionada con `amor`, aunque esa
#   palabra no esta en el texto por lo que para este caso puede funcionar.
# Otra conjutación `lovable` no se encontraría con ese regex,
#   pero tampoco esta en el texto
# Esto da a entender que una regex quizá no es lo mejor para detectar variantes.
# Se consultaron la lista de palabras que comienzan con `lov` para tener una
#   mejor idea en:
#   https://www.merriam-webster.com/wordfinder/classic/begins/all/-1/lov/1

results = [res for doc in docs if (res := re.findall(r"\blove\w*|\bloving\w*", doc, re.IGNORECASE))]
```

```python
np_res = np.concatenate(results)
print(f"total variantes de love: {len(np_res)}")
np_res
```

*   **Pregunta 8.**  

Busca e imprime todas las palabras, variantes de "so" y "good", que tengan dos o más "o" en "so" y 3 o más "o" en good.

Indica cuántas encontraste.


```python
# Usamos:
#   o{2,}, para detectar 2 o mas ocurrencias de la letra 'o' en so
#   o{3,}, para detectar 3 o más ocurrencias de la letra 'o; en good
#   \d+ por si escriben varias 'd' al final de `gooodd`
results = [res for doc in docs if (res := re.findall(r"\bso{2,}\b|\bgo{3,}d+", doc, re.IGNORECASE))]
```

*   **Pregunta 9.**  

Busca e imprime todas las palabras que tengan una longitud mayor estrictamente a 10 caracteres alfabéticos.

No se consideran los signos de puntuación o caracteres especiales en la longitud de estas cadenas, solo caracteres alfabéticos en mayúsculas o minúsculas.

```python
# Usamos el flag de IGNORECASE para simplificar la expresión ([a-zA-Z])
#  Basicamente se revisa que la palabra completa sea de al menos 11 caracteres
#  a-z sin importar si es mayúscula o minúscula.

results = [res for doc in docs if (res := re.findall(r"\b[a-z]{11,}\b", doc, re.IGNORECASE))]
```

*   **Pregunta 10.**  

Busca e imprime todas las palabras que inician con una letra mayúscula y terminan con una minúscula, pero que además no sea la primera palabra del comentario/string.

Indica la cantidad de resultados obtenidos.

```python
results = []
for doc in docs:
  # Buscamos palabras completas (\b)
  #   que empiecen con mayúscula [A-Z]
  #   pueden seguir cualquier cantidad de letras mayúsculas o minúsculas [a-zA-Z]
  #   debe terminar en minúscula [a-z]
  all_valid_words = re.findall(r"\b[A-Z][a-zA-Z]*[a-z]\b", doc)
  # Primer palabra del string que empieza con Mayúscula y termina con minúscula
  #  Excluimos cualquier carácter que no sea letra del inicio del string
  first_word = re.findall(r'^[^a-zA-Z]*([A-Z][a-zA-Z]*[a-z]\b)', doc)
  # Si la primer palabra de la oración cumple el criterio, la quitamos
  if first_word and first_word[0] == all_valid_words[0]:
      results += all_valid_words[1:]
  else:
    results += all_valid_words
```

```python
print(f"total de palabras: {len(results)}")
# convertimos a numpy array para que se impriman más elementos por línea
np.array(results)
```

#**Parte 3. Proceso de limpieza.**

*   **Pregunta 13.**  

Ahora realiza un proceso de limpieza del corpus que incluya los siguientes procesos:

*   Solo se deben considerar caracteres alfabéticos. Es decir, se eliminan todos los signos de puntuación y caracteres especiales.
*   Todos los caracteres alfabéticos se transforman a minúsculas.
*   Se deben eliminar todos los espacios en blanco adicionales que se puedan encontrar en cada comentario.

Al finalizar dicho proceso de limpieza, imprime el resultado de los primeros 10 comentarios resultantes.
   




```python
def clean_corpus(text):
  # transformar a minúsculas
  text = text.lower()

  # separamos por espacio
  tokens = text.split()

  # Eliminar todo lo que NO sea alfabético
  tokens = [re.sub(r'[^\w]+', '', token) for token in tokens]

  # quitamos tokens no alfabéticos
  tokens = [w for w in tokens if w.isalpha()]

  # Eliminamos espacios en blanco, si es que aún existen
  tokens = [re.sub(r'\s+', ' ', token) for token in tokens]
  return tokens

corpus_tokenizado = [clean_corpus(doc) for doc in docs]
```

```python
for i in range(10):
  print(corpus_tokenizado[i])
```

Al terminar calcula el total de tokens obtenido en todo el corpus.

```python
# Sumamos la cantidad de tokens en cada documento
total_tokens = sum(len(tokens) for tokens in corpus_tokenizado)

print(f"Total de tokens en el corpus: {total_tokens}")
```
